```python
# W7A1
# Viết hàm tìm kiếm nhị phân trên một danh sách đã được sắp xếp tăng dần. 
# Trả về chỉ số của phần tử hoặc -1 nếu không tìm thấy.
def binary_search(arr,target):
    left = 0
    right = len(arr) - 1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        elif target < arr[mid]:
            right = mid - 1
        else:
            left = mid + 1
    return -1
arr = list(map(int,input().split()))
target = int(input())
print(binary_search(arr,target))
            
            
```

```python
# W7A2
# Cho danh sách arr và một số nguyên x, đếm xem x xuất hiện bao nhiêu lần trong danh sách

def count_occurrences(arr, x):
    count = 0
    for i in arr:
        if i == x:
            count += 1
    return count
arr = list(map(int,input().split()))
x = int(input())
print(count_occurrences(arr,x))
```

```python
# W7A3
# Viết hàm thực hiện thuật toán bubble sort, Selection Sort để sắp xếp danh sách 
# theo thứ tự tăng dần. Sắp xếp nổi bọt (Bubble Sort) Sắp xếp chọn (Selection Sort)

def selection_sort(arr):
    for i in range (len(arr)):
        min_idx = i
        for j in range (i + 1,len(arr)):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[min_idx],arr[i] = arr[i],arr[min_idx]       
    return arr
arr = list(map(int,input().split()))
print(*selection_sort(arr))
                
            
```

```python
# W7A4
# Viết hàm tìm phần tử có số lần xuất hiện nhiều nhất trong danh sách. Nếu có 
# nhiều phần tử cùng số lần, trả về phần tử xuất hiện trước

def count_element(arr):
    dic = {}
    for i in range(len(arr)):
        if arr[i] not in dic:
            dic[arr[i]] = 1
        else:
            dic[arr[i]] += 1
    max_value = 0
    for key in dic:
        if dic[key] > max_value:
            max_value = dic[key]
            max_key = key
    return f"{max_key} xuat hien nhieu nhat, som nhat, {max_value} lan"
             
arr = list(map(int,input().split()))
print(count_element(arr))
        
```

```python
# W7A8
# Bạn được cung cấp: một danh sách các khoảng thời gian interval = [[l1, r1], [l2, r2], ...] và một danh sách 
# các truy vấn queries = [q1, q2, ...] Với mỗi truy vấn q, hãy tìm độ dài nhỏ nhất của khoảng thời gian [li, ri] sao 
# cho li ≤ q ≤ ri. Nếu không tồn tại khoảng thời gian nào như vậy, hãy trả về -1 cho truy vấn đó. solve in < O(m.n), hint: 
# binary search in the queries list

def check(arr,lst):
    res = []
    for i in range(len(lst)):
        temp = []
        for j in range(len(arr)):
            if lst[i] >= arr[j][0] and lst[i] <= arr[j][1]:
                temp.append(arr[j][1] - arr[j][0] + 1)
        if len(temp) == 0:
            res.append(-1)
        else:
            res.append(min(temp))
    return res
arr = [[5, 10], [2, 3], [8, 8]]
lst = [1, 3, 5, 8, 9]
print(check(arr,lst))
            
            
```

```python
# W7A9
# Cho một danh sách các số nums, với mỗi phần tử nums[i], hãy đếm xem có bao nhiêu phần tử bên phải 
# nhỏ hơn nums[i]. Trả về danh sách các số đếm. Efficient solution: using heap or binary search tree

def check(nums):
    res = []
    for i in range(len(nums)):
        count = 0
        for j in range(i,len(nums)):
            if nums[i] > nums[j]:
                count += 1
        res.append(count)
    return res
nums = list(map(int,input().split()))
print(check(nums))
            
```

```python
# W7A10
# Cho một danh sách các số nguyên nums và một số nguyên m, hãy tìm tổng mảng con lớn nhất theo 
# modulo m. Tức là, với bất kỳ mảng con nums[i:j] nào, hãy tính sum(nums[i:j]) % m và trả về kết quả lớn nhất có thể.
# Gợi ý: Duy trì tổng tiền tố (prefix sum) mod m và sử dụng tìm kiếm nhị phân để tìm tổng tiền tố nhỏ nhất chỉ 
# lớn hơn tổng hiện tại một chút (để tối đa hóa hiệu số bao quanh)

def quan(nums,m):
    res = []
    for i in range (len(nums)):
        for j in range (i,len(nums) + 1):
            res.append(sum(nums[i:j]) % m)
    return max(res)
nums = [7, 11, 14, 2, 8, 4]
m = 12
print(quan(nums,m))
    
```

```python
# W7A11
# Các trình duyệt web thường có một chức năng tìm kiếm một từ trong một trang web. Chẳng hạn khi 
# bạn tìm kiếm từ "UET" trên trang UETCodeHub, trình duyệt sẽ hiện ra có một từ "UET". Hãy viết một chương trình để xác định 
# có bao nhiêu từ xuất hiện trong một câu. Đầu vào: Một chuỗi kí tự trên một dòng. Từ khóa cần tìm ở dòng tiếp theo.
# Đầu ra: Số lần xuất hiện của từ tìm kiếm.
def quan(string,key_word):
    count = 0 
    for i in range(len(string)):
        if string[i:i + len(key_word)] == key_word:
            count += 1
    return count
string = str(input())
key_word = str(input())
print(quan(string,key_word))
```
