```python
# W11A1
# """Bất cứ một số nguyên dương nào cũng tồn tại một số nguyên dương là số đảo ngược của nó.
# Hãy viết hàm def reverse (path: str) -> int: để tìm số nguyên đảo ngược của số n 
# Đầu vào 
# Đầu vào của hàm là một xâu biểu diễn đường dẫn đến tệp. Nội dung tệp chứa duy nhất một số nguyên dương. 
# Đầu ra 
# Hàm trả về số nguyên dương đảo ngược của số trong tệp đầu vào."""

def reverse(path):
    with open (path, 'r') as f:
        n_str = f.read().strip()
    n_str = n_str[::-1]
    return int(n_str)
result = reverse("minhbeta.txt")
print(result)
    
```

```python
# W11A2
# Bạn đang chơi trò chơi đánh cắp với một người bạn, luật chơi như sau: 
# Có một đống đá trên bàn, trong mỗi lượt chơi, các bạn có thể loại bỏ từ 1 đến 3 hòn đá.
# Người loại bỏ hòn đá cuối cùng sẽ là người chiến thắng. Bạn sẽ là người được ưu tiên là người chơi trước
# Cả hai bạn đều rất thông minh và có chiến lược tối ưu cho trò chơi. 
# Viết hàm def canWinNim (path: str) -> bool để xác định xem bạn có thể 
# giành chiến thắng trong trò chơi với số lượng đá đã biết trước không. 
# Gợi ý: nếu trong đống đá có 4 hòn đá, thì bạn sẽ không bao giờ thắng 
# được trò chơi: dù bạn lấy 1,2 hay 3 hòn đá, thì bạn của bạn sẽ luôn là người chiến thắng. 
# Đầu vào 
# Đầu vào của hàm là một xâu biểu diễn đường dẫn đến tệp. Nội dung tệp lưu duy nhất một số nguyên là số lượng đá của trò chơi. 
# Đầu ra 
# Hàm trả về true nếu bạn có thể thắng, ngược lại trả về false.

def canwinnim(path):
    with open (path, 'r') as f:
        n_str = int(f.read().strip())
        return n_str % 4 != 0
result = canwinnim('minhbeta.txt')
print(result)
```

```python
# W11A4
# Harry Potter là tên của bộ truyện (gồm bảy phần) của nữ nhà văn nước Anh J. K. Rowling. 
# Bộ truyện viết về những cuộc phiêu lưu phù thủy của cậu bé Harry Potter và những người bạn
# Ronald Weasley, Hermione Granger, lấy bối cảnh tại Trường Phù thủy và Pháp sư Hogwarts nước Anh.
# Những cuộc phiêu lưu tập trung vào cuộc chiến của Harry Potter trong việc chống lại tên Chúa tể
# hắc ám Voldemort - người có tham vọng muốn trở nên bất tử, thống trị thế giới phù thủy, nô dịch hóa
# những người phi pháp thuật và tiêu diệt những ai cản đường hắn đặc biệt là Harry Potter.
# Trong bộ truyện, người ta kể lại rằng Tom Riddle, một chàng trai dị thường người mà sau 
# đó trở thành phù thủy đáng sợ nhất mọi thời đại, đã sáng tạo ra biệt danh xấu xa của mình 
# như một kiểu đảo chữ (anagram). Tom Marvolo Riddle trở thành "I Am Lord Voldemort.". 
# Hai chuỗi là anagram của nhau nếu các chữ cái của chuỗi này có thể được sắp xếp để tạo 
# thành chuỗi kia. Hãy viết hàm def areAnagrams(path1: str, path2: str) trả về true nếu hai 
# chuỗi đầu vào là anagram của nhau ngược lại false. Hàm không phân biệt chữ hoa chữ thường và 
# không quan tâm đến các dấu câu và dấu cách.
# Đầu vào là 2 tên file trên 2 dòng, mỗi file chứa một chuỗi kí tự.
# Đầu ra: là True nếu 2 chuỗi kí tự là anagram của nhau, False nếu ngược lại

def areAnagrams(path1: str, path2: str) -> bool:
    def normalize(s: str) -> str:
        return ''.join(c for c in s.lower() if c.isalpha())
    with open(path1, 'r', encoding='utf-8') as f1:
        s1 = f1.read().strip()
    with open(path2, 'r', encoding='utf-8') as f2:
        s2 = f2.read().strip()
    s1 = normalize(s1)
    s2 = normalize(s2)
    return sorted(s1) == sorted(s2)
print(areAnagrams('minhbot.txt','minhbeta.txt'))
```

```python
# W11A10

def get_money(path):
    with open(path, 'r', encoding = 'utf-8') as f:
        data = f.read().strip()
    return list(map(int,data.split()))
def rob(houses):
    if len(houses) == 0:
        return 0
    if len(houses) == 1:
        return houses[0]
    prev2 = houses[0]               
    prev1 = max(houses[0], houses[1])  
    for i in range(2, len(houses)):
        current = max(prev1, prev2 + houses[i])
        prev2 = prev1
        prev1 = current
    return prev1
houses = get_money("minhbeta.txt")
print(rob(houses))

    
```

```python
# W11A12
def maximumProduct(path: str) -> int:
    with open(path, 'r', encoding='utf-8') as f:
        n = int(f.readline().strip())
        nums = list(map(int, f.readline().split()))
    nums.sort()
    product1 = nums[-1] * nums[-2] * nums[-3]   
    product2 = nums[0] * nums[1] * nums[-1]     
    return max(product1, product2)
path = input().strip()
print(maximumProduct(path))
```

```python
# W11A13
def moveZeroes(fileName):
    with open(fileName, 'r', encoding='utf-8') as f:
        n = int(f.readline().strip())
        nums = list(map(int, f.readline().split()))
    result = []
    zero_count = 0
    for num in nums:
        if num == 0:
            zero_count += 1
        else:
            result.append(num)
    result.extend([0] * zero_count)
    return result
fileName = input().strip()
print(moveZeroes(fileName))
```

```python
# W11A14
def averageTime(path):
    total_time = 0.0
    count = 0
    with open(path, 'r', encoding='utf-8') as f:
        lines = f.readlines()
    for line in lines[1:]:
        data = line.strip().split(',')
        time = float(data[1])
        total_time += time
        count += 1
    average = total_time / count
    return round(average, 2)
print(averageTime("minhbeta.txt"))
```

```python
# W11A15
def findMovies(path: str) -> list[str]:
    result = []
    with open(path, 'r', encoding='utf-8') as f:
        lines = f.readlines()
    for line in lines[1:]:
        data = line.strip().split(',')
        movie = data[1].strip()
        description = data[2].strip().lower()
        rating = float(data[3])
        if rating > 8.0 and description != "boring":
            result.append(movie)
    if not result:
        return ["All are boring!"]
    return result
print(findMovies('minhbeta.txt'))
```

```python
# W11A16
import math

def printPerimeter(path: str):
    with open(path, 'r', encoding='utf-8') as f:
        lines = f.readlines()
    i = 0
    while i < len(lines):
        line = lines[i].strip()
        if line.startswith("TYPE"):
            shape_type = line.split(":")[1].strip()
            info = lines[i + 1].strip()
            if shape_type == "SQUARE":
                side = float(info.split(":")[1])
                perimeter = 4 * side
            elif shape_type == "RECTANGLE":
                parts = info.split()
                width = float(parts[1])
                height = float(parts[3])
                perimeter = 2 * (width + height)
            elif shape_type == "CIRCLE":
                radius = float(info.split(":")[1])
                perimeter = 2 * math.pi * radius
            print(f"{perimeter:.2f}")
            i += 2  
        else:
            i += 1 
printPerimeter('minhbeta.txt')
```

```python
# W11A17
def productExceptSelf(path: str) -> list[int]:
    with open(path, 'r', encoding='utf-8') as f:
        numbers = list(map(int, f.read().split()))
    n = len(numbers)
    result = []
    for i in range(n):
        product = 1
        for j in range(n):
            if i != j:
                product *= numbers[j]
        result.append(product)
    print(result)
    return result
productExceptSelf('minhbeta.txt')
```

```python
# W11A18
import math

def estimatedTime(path: str):
    with open(path, 'r', encoding='utf-8') as f:
        lines = f.readlines()
    src_data = lines[0].split()
    x1 = float(src_data[2])
    y1 = float(src_data[4])
    des_data = lines[1].split()
    x2 = float(des_data[2])
    y2 = float(des_data[4])
    velocity_data = lines[2].split()
    velocity = float(velocity_data[1])
    distance = math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)
    time = distance / velocity
    return round(time, 2)
print(estimatedTime("minhbeta.txt"))

```

```python
# W11A20
def imageSmoother(path: str):
    with open(path, 'r', encoding='utf-8') as f:
        M = int(f.readline().strip())
        image = []
        for _ in range(M):
            image.append(list(map(int, f.readline().split())))
    result = [[0] * M for _ in range(M)]
    directions = [-1, 0, 1]
    for i in range(M):
        for j in range(M):
            total = 0
            count = 0
            for dx in directions:
                for dy in directions:
                    ni = i + dx
                    nj = j + dy
                    if 0 <= ni < M and 0 <= nj < M:
                        total += image[ni][nj]
                        count += 1
            result[i][j] = total // count
    for row in result:
        print(*row)
imageSmoother("minhbeta.txt")
```
